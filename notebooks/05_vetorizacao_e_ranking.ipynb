# 05 - Vetorização (CountVectorizer e TF-IDF) e Ranking de Termos

Este notebook carrega o corpus pré-processado e as stopwords curadas, gera as matrizes de representação textual via `CountVectorizer` e `TfidfVectorizer` e exporta os rankings de termos em `results/`.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.curadoria import carregar_stopwords_curadas
from src.vetorizacao import carregar_corpus, vetorizar_count, vetorizar_tfidf
```

## 1. Parâmetros de Vetorização

```python
INPUT_PATH = PROJECT_ROOT / "data" / "processed" / "dataset_preprocessado.csv"
CURADORIA_FILE = PROJECT_ROOT / "data" / "processed" / "curadoria_stopwords.csv"
RESULTS_DIR = PROJECT_ROOT / "results"

COLUMN_ID = "id"
COLUMN_TEXT = "clean_text"

NGRAM_RANGE = (1, 1)
MAX_DF = 0.85
MIN_DF = 5
```

## 2. Carregamento do Corpus e Stopwords

```python
df_corpus = carregar_corpus(INPUT_PATH, column_id=COLUMN_ID, column_text=COLUMN_TEXT)

if CURADORIA_FILE.exists():
    stopwords_final = carregar_stopwords_curadas(CURADORIA_FILE)
else:
    print("Arquivo de curadoria não encontrado; prosseguindo sem stopwords adicionais.")
    stopwords_final = set()
```

## 3. Vetorização por CountVectorizer e Ranking por Frequência

```python
count_vectorizer, countv_matrix, ranking_countv = vetorizar_count(
    df_corpus[COLUMN_TEXT],
    stop_words=stopwords_final,
    ngram_range=NGRAM_RANGE,
    max_df=MAX_DF,
    min_df=MIN_DF,
)

RESULTS_DIR.mkdir(parents=True, exist_ok=True)
ranking_countv.to_csv(RESULTS_DIR / "ranking_countvectorizer.csv", index=False, encoding="utf-8")
ranking_countv.head(20)
```

## 4. Vetorização por TF-IDF e Ranking por Soma dos Scores

```python
tfidf_vectorizer, tfidf_matrix, ranking_tfidf = vetorizar_tfidf(
    df_corpus[COLUMN_TEXT],
    stop_words=stopwords_final,
    ngram_range=NGRAM_RANGE,
    max_df=MAX_DF,
    min_df=MIN_DF,
)

ranking_tfidf.to_csv(RESULTS_DIR / "ranking_tfidf.csv", index=False, encoding="utf-8")
ranking_tfidf.head(20)
```
