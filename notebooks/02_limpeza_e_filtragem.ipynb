# 02 - Limpeza e Filtragem de Qualidade

Este notebook carrega o dataset bruto de `data/raw/`, remove resumos vazios, tags HTML/JATS, textos com mojibake, artigos fora do idioma alvo e transcrições de vídeo, salvando o resultado em `data/processed/`.

```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.limpeza import limpar_dataset
```

## 1. Parâmetros de Entrada e Saída

```python
INPUT_FILE = PROJECT_ROOT / "data" / "raw" / "dataset_bruto.csv"
OUTPUT_FILE = PROJECT_ROOT / "data" / "processed" / "dataset_limpo.csv"
TARGET_LANG = "en"
```

## 2. Execução do Pipeline de Limpeza

```python
df_init = pd.read_csv(INPUT_FILE, encoding="utf-8")
df_clean = limpar_dataset(df_init, target_lang=TARGET_LANG)

OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)
df_clean.to_csv(OUTPUT_FILE, index=False, encoding="utf-8")
print(f"Dataset limpo salvo em: {OUTPUT_FILE}")
```
