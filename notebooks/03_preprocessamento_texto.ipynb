# 03 - Pré-processamento Linguístico (Lematização e Stopwords)

Este notebook carrega o dataset limpo, aplica lematização e remoção de stopwords com spaCy e exporta o corpus pronto para modelagem em `data/processed/`.

```python
import sys
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.preprocessamento import carregar_modelo_spacy, preprocessar_dataset
```

## 1. Parâmetros e Stopwords Customizadas

```python
INPUT_FILE = PROJECT_ROOT / "data" / "processed" / "dataset_limpo.csv"
OUTPUT_FILE = PROJECT_ROOT / "data" / "processed" / "dataset_preprocessado.csv"

SPACY_MODEL = "en_core_web_sm"
STOPWORDS_EXTRA = {"palavra1", "palavra2"}
```

## 2. Execução do Pré-processamento

```python
df_clean = pd.read_csv(INPUT_FILE, encoding="utf-8")
nlp = carregar_modelo_spacy(SPACY_MODEL)

df_custom = preprocessar_dataset(df_clean, nlp, stopwords_extra=STOPWORDS_EXTRA)

OUTPUT_FILE.parent.mkdir(parents=True, exist_ok=True)
df_custom.to_csv(OUTPUT_FILE, index=False, encoding="utf-8")
print(f"Dataset pré-processado salvo em: {OUTPUT_FILE}")
```
