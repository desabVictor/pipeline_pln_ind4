```python
# Reconstrói os resumos invertidos pela API
def rebuild_abstract(inverted):
    if not inverted or not isinstance(inverted, dict):
        return ""
    # Cria um mapa de posição:palavra e ordena pelas chaves (posições)
    pos_map = {pos: word for word, positions in inverted.items() for pos in positions}
    return " ".join(pos_map[i] for i in sorted(pos_map.keys()))

print("Processando dados e reconstruindo abstracts...")

# Lista para o Dataset
data_list = []
for w_id, work in unique_works.items():
    data_list.append({
        "id": w_id,
        "title": work.get("title"),
        "year": work.get("publication_year"),
        "doi": work.get("doi"),
        "cited_by": work.get("cited_by_count"),
        "abstract": rebuild_abstract(work.get("abstract_inverted_index")),
        # Une as categorias encontradas em uma única string separada por vírgula
        "all_areas": ", ".join(sorted(list(id_to_categories[w_id])))
    })

# Criação do DataFrame e salvamento em CSV
NAME_FILE = "o-nome-do-arquivo-csv"
df_final = pd.DataFrame(data_list)
df_final.to_csv(f"{NAME_FILE}.csv", index=False, encoding='utf-8')

print(f"--- SUCESSO ---")
print(f"Total de artigos processados: {len(df_final)}")
print(f"Arquivo '{NAME_FILE}.csv' salvo com sucesso!")
```
