# Carregamento do dataframe

```python
INPUT_PATH = "insira-o-nome-do-arquivo-csv"
COLUMN_ID = "id"
COLUMN_TEXT = "clean_text"

df_corpus = pd.read_csv(INPUT_PATH, encoding='utf-8')

assert COLUMN_ID in df_corpus.columns, (
    f"Coluna '{COLUMN_ID}' não encontrada. Colunas disponíveis: "
    f"{list(df_corpus.columns)}"
)
assert COLUMN_TEXTO in df_corpus.columns, (
    f"Coluna '{COLUMN_TEXT}' não encontrada. Colunas disponíveis: "
    f"{list(df_corpus.columns)}"
)

df_corpus = df_corpus.dropna(subset=[COLUMN_TEXT]).reset_index(drop=True)
print(f"Total de documentos carregados: {len(df_corpus)}")
```

# Vetorização do dataframe por CountVectorizer

```python
# Parâmetros do nosso CountVectorizer
count_vectorizer = CountVectorizer(
    stop_words=list(stopwords_final),
    ngram_range=(1,1),
    max_df=0.85,
    min_df=5
)

# Vetorizando nosso dataset
countv_matrix = count_vectorizer.fit_transform(df_corpus[COLUM_TEXT])
feature_names_countv = count_vectorizer.get_feature_names_out()

print(f"Dimensão da matriz CountVectorizer (documentos x termos): {countv_matrix.shape}")
print(f"Tamanho do vocabulário: {len(feature_names_countv)}")
```

# Ranking de termos por frequência

```python
countv_scores = np.asarray(countv_matrix.sum(axis=0)).flatten()

ranking_countv = (
    pd.DataFrame({"term": feature_names_countv, "countv_scores": countv_scores})
    .sort_values("countv_scores", ascending=False)
    .reset_index(drop=True)
)
```
