# Carregamento do dataframe

```python
INPUT_PATH = "insira-o-nome-do-arquivo-csv"
COLUMN_ID = "id"
COLUMN_TEXT = "clean_text"

df_corpus = pd.read_csv(INPUT_PATH, encoding='utf-8')

assert COLUMN_ID in df_corpus.columns, (
    f"Coluna '{COLUMN_ID}' não encontrada. Colunas disponíveis: "
    f"{list(df_corpus.columns)}"
)
assert COLUMN_TEXTO in df_corpus.columns, (
    f"Coluna '{COLUMN_TEXT}' não encontrada. Colunas disponíveis: "
    f"{list(df_corpus.columns)}"
)

df_corpus = df_corpus.dropna(subset=[COLUMN_TEXT]).reset_index(drop=True)
print(f"Total de documentos carregados: {len(df_corpus)}")
```

# Vetorização do dataframe por TfidfVectorizer

```python
print("\nIniciando Extração de Conhecimento (TF-IDF)...")

vectorizer = TfidfVectorizer(stop_words=list(stopwords_final), ngram_range=(1,1), max_df=0.85, min_df=5)
tfidf_matrix = vectorizer.fit_transform(df_corpus[COLUMN_TEXT])
feature_names = vectorizer.get_feature_names_out()

print("Extração de conhecimento concluída!")
print(f"Dimensão da matriz TF-IDF (documentos x termos): {tfidf_matrix.shape}")
print(f"Tamanho do vocabulário: {len(feature_names)}")
```

# Ranking de termos por soma

```python
sum_scores = np.asarray(tfidf_matrix.sum(axis=0)).flatten()

ranking_sum = (
    pd.DataFrame({"term": feature_names, "sum_tfidf": soma_scores})
    .sort_values("sum_tfidf", ascending=False)
    .reset_index(drop=True)
)
```
